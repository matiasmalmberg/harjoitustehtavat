```python
import random

thrownDiceNumbers = [0, 0, 0, 0, 0]

for i in range(5):
    thrownDiceNumbers[i] = random.randint(1, 6)

print(thrownDiceNumbers)

total = sum(thrownDiceNumbers)

print(total)

print(max(thrownDiceNumbers))
```
